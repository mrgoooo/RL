```python

import time

# 1. Tworzymy środowisko z trybem renderowania 'human', żeby widzieć je w konsoli
env = gym.make('Taxi-v3', render_mode = 'ansi')

# 2. Resetujemy środowisko do stanu początkowego
state, info = env.reset()

try:
    for step in range(20):  # Wykonamy 20 losowych kroków
        # 3. Wybieramy losową akcję (0-5)
        # Akcje: 0=dół, 1=góra, 2=prawo, 3=lewo, 4=pasażer wsiada, 5=pasażer wysiada
        action = env.action_space.sample()

        # 4. Wykonujemy akcję w środowisku
        observation, reward, terminated, truncated, info = env.step(action)

        # 5. Wyświetlamy informację o kroku
        env.render()
        

        # Jeśli taksówka osiągnęła cel lub czas się skończył
        if terminated or truncated:
            print("Koniec epizodu!")
            state, info = env.reset()
        
        # Krótka pauza, żebyś mógł zobaczyć co się dzieje
        time.sleep(0.5)

finally:
    # 6. Zamykamy środowisko po zakończeniu pętli
    env.close()
```
